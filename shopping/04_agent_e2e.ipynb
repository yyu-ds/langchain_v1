## Agents

```python
#shopping_order_agent = agents.orders_agent_setup()
orchestrator_agent = agents.orchestrator_agent_setup() 

agent = orchestrator_agent
```

```python


def chat_with_agent(message, history):
    """
    Handle chat messages with the shopping agent.
    
    Args:
        message: User's input message
        history: Chat history
    
    Returns:
        Assistant's reply
    """
    if not message.strip():
        return ""
    
    # Use a fixed thread_id for the session
    config = {"configurable": {"thread_id": "user_session_chat"}}
    
    try:
        # Create a proper HumanMessage object
        resp = agent.invoke(
            {"messages": [HumanMessage(content=message)]}, 
            config=config
        )
        reply = resp["messages"][-1].content
    except Exception as e:
        reply = f"Error: {e}"
    
    return reply

# Create Gradio interface using ChatInterface
demo = gr.ChatInterface(
    fn=chat_with_agent,
    title="Shopping Assistant",
    description="Ask questions about your orders, spending, and purchases.",
    chatbot=gr.Chatbot(height=500),
    textbox=gr.Textbox(placeholder="Type your message here...", lines=1),
    examples=[
        "How many orders did I make?",
        "What was my last order?",
        "Tell me about item 2009728644?",
        "What should i give for mothers day?"
    ]
)

# Launch the interface
demo.launch(share=False, debug=True)
```
